```python
# 新任务：量化所有掩膜并保存为新的 .pt 文件
import torch
import numpy as np
from typing import Union

def quantize_all_masks_and_save(
    input_path: str,
    output_path: str,
    bits: int = 5,
    apply_background_shift: bool = True,  # 是否将背景从 0 平移到 -π
    save_as_phase: bool = True,
):
    """
    加载 3D 掩膜张量，逐个量化（周期性处理），保存为新的 .pt 文件
    
    参数:
        input_path: 输入 .pt 文件路径
        output_path: 输出 .pt 文件路径
        bits: 量化位数（默认 5 位，32 级）
        apply_background_shift: 是否将背景从 0 平移到 -π（制造优化）
        save_as_phase: True 保存为相位值 [-π,π]，False 保存为整数 [0,31]
    """
    # 加载原始数据
    obj = torch.load(input_path, map_location="cpu")
    a_torch = obj.detach().cpu()  # 保持 torch 格式
    print(f"[Step 1] 加载数据")
    print(f"  形状: {a_torch.shape}")
    print(f"  原始范围: [{torch.min(a_torch).item():.4f}, {torch.max(a_torch).item():.4f}]")
    
    # 🔧 背景平移（制造优化）：0 → -π
    if apply_background_shift:
        print(f"\n[Step 2] 背景平移优化（0 → -π）")
        a_torch = a_torch + np.pi  # 整体平移 +π
        print(f"  平移 +π 后范围: [{torch.min(a_torch).item():.4f}, {torch.max(a_torch).item():.4f}]")
    else:
        print(f"\n[Step 2] 跳过背景平移")
    
    # 周期性包裹到 [-π, π]
    print(f"\n[Step 3] 周期性包裹")
    wrapped = torch.remainder(a_torch + np.pi, 2 * np.pi) - np.pi
    print(f"  包裹后范围: [{torch.min(wrapped).item():.4f}, {torch.max(wrapped).item():.4f}]")
    
    # 量化
    levels = 1 << bits  # 2^bits
    step = (2 * np.pi) / float(levels)
    
    print(f"\n[Step 4] 量化到 {levels} 级")
    print(f"  量化步长: {step:.6f} rad ({step*180/np.pi:.3f}°)")
    
    # 计算整数级别索引（用于制造）
    indices = torch.round((wrapped + np.pi) / step).to(torch.int16)
    indices = torch.clamp(indices, 0, levels - 1)
    
    # 量化后的相位值
    quantized_phase = torch.round(wrapped / step) * step
    
    # 统计信息
    unique_indices = torch.unique(indices).numel()
    print(f"  实际唯一级别数: {unique_indices}/{levels}")
    print(f"  整数级别范围: [{torch.min(indices).item()}, {torch.max(indices).item()}] (用于二进制刻蚀)")
    print(f"  量化相位范围: [{torch.min(quantized_phase).item():.4f}, {torch.max(quantized_phase).item():.4f}] rad")
    
    # 显示级别分布直方图（简化版）
    unique_vals, counts = torch.unique(indices, return_counts=True)
    print(f"\n  级别分布 (前10个):")
    for i in range(min(10, len(unique_vals))):
        idx = unique_vals[i].item()
        phase_val = (idx / (levels - 1)) * (2 * np.pi) - np.pi
        print(f"    级别 {idx:2d} → 相位 {phase_val:+.4f} rad: {counts[i].item():7d} 像素")
    if len(unique_vals) > 10:
        print(f"    ... (还有 {len(unique_vals)-10} 个级别)")
    
    # 选择保存格式
    if not save_as_phase:
        quantized_tensor = indices
        print(f"\n[Step 5] 保存为整数索引")
        print(f"  数据类型: {quantized_tensor.dtype}")
    else:
        quantized_tensor = quantized_phase
        print(f"\n[Step 5] 保存为相位值")
        print(f"  数据类型: {quantized_tensor.dtype}")
    
    # 保存
    torch.save(quantized_tensor, output_path)
    print(f"\n✅ 已保存到: {output_path}")
    print(f"   形状: {quantized_tensor.shape}")
    
    return quantized_tensor


# 执行量化并保存（保存为相位值，使用与 quantize_masks 一致的算法）
input_path = r"d:\zlw\project\MPLC_CUDA\results_2\masks_full.pt"
output_path = r"d:\zlw\project\MPLC_CUDA\results_2\masks_full_quantized_5bit_phase.pt"

print("="*60)
print("量化任务配置:")
print(f"  输入: {input_path}")
print(f"  输出: {output_path}")
print(f"  量化位数: 5 bits (32 levels)")
print(f"  背景平移: True")
print("="*60)

quantized_tensor = quantize_all_masks_and_save(
    input_path=input_path,
    output_path=output_path,
    bits=5,
    apply_background_shift=True,
    save_as_phase=True,
)
```

```python
#反射损耗
import numpy as np

def calculate_loss_db(efficiency: float, num_stages: int = 1) -> tuple:
    """
    计算光学损耗
    
    参数:
        efficiency: 单次效率 (0-1)
        num_stages: 级联级数
    
    返回:
        (单次损耗dB, 总损耗dB, 总效率)
    """
    single_loss_db = -10 * np.log10(efficiency)
    total_loss_db = single_loss_db * num_stages
    total_efficiency = 10 ** (-total_loss_db / 10)
    return single_loss_db, total_loss_db, total_efficiency

# ============================================================
# 损耗计算示例
# ============================================================

print("=" * 60)
print("MPLC 系统损耗分析")
print("=" * 60)

# 场景 1: 单层效率 98%，9 层级联
eff1 = 0.98
stages1 = 9
loss1, total_loss1, final_eff1 = calculate_loss_db(eff1, stages1)

print(f"\n[场景 1] 单层效率 {eff1:.1%}，{stages1} 层级联")
print(f"  单层损耗: {loss1:.6f} dB")
print(f"  总损耗:   {total_loss1:.3f} dB")
print(f"  总效率:   {final_eff1:.4%}")

# 场景 2: 单层效率 99%，9 层级联
eff2 = 0.99
stages2 = 9
loss2, total_loss2, final_eff2 = calculate_loss_db(eff2, stages2)

print(f"\n[场景 2] 单层效率 {eff2:.1%}，{stages2} 层级联")
print(f"  单层损耗: {loss2:.6f} dB")
print(f"  总损耗:   {total_loss2:.3f} dB")
print(f"  总效率:   {final_eff2:.4%}")

# 场景 3: 混合损耗（98% × 9层 + 99% × 9层）
print(f"\n[场景 3] 混合系统")
print(f"  子系统 A: {eff1:.1%} × {stages1} 层 = {total_loss1:.3f} dB")
print(f"  子系统 B: {eff2:.1%} × {stages2} 层 = {total_loss2:.3f} dB")
combined_loss = total_loss1 + total_loss2
combined_eff = 10 ** (-combined_loss / 10)
print(f"  总损耗:   {combined_loss:.3f} dB")
print(f"  总效率:   {combined_eff:.4%}")


```

```python
# 检查频率轴是否 off-by-one（半格偏移）以及 meshgrid 维度是否与 (Ny, Nx) 对齐
import math
import torch


def check_off_by_one_and_meshgrid(Nx=None, Ny=None, dx=None):
    # 1) 自动获取配置（若可），否则用常见默认值
    if Nx is None or Ny is None or dx is None:
        try:
        #     import MPLC_CUDA2 as m2
        #     cfg = m2.parse_cfg()
        #     Nx = Nx or int(cfg.Nx)
        #     Ny = Ny or int(cfg.Ny)
        #     dx = dx or float(cfg.pixelSize)
        # except Exception:
        #     # 兜底默认：与项目当前默认一致
            Nx = Nx or 9
            Ny = Ny or 6
            dx = dx or 1e-6
        except Exception:
            # 兜底默认：与项目当前默认一致
            1

    print("==== Sampling summary ====")
    print(f"Nx={Nx}, Ny={Ny}, dx={dx:g} m")

    device = "cpu"
    dtype = torch.float64

    # 2) 我们当前写法（基于中心化索引 -(N-1)/2 .. (N-1)/2）

    nx_t = (torch.arange(Nx, dtype=dtype, device=device) - Nx // 2) 
    ny_t = (torch.arange(Ny, dtype=dtype, device=device) - Ny // 2) 
    print(torch.arange(Nx, dtype=dtype, device=device))
    print("Nx//2:", Nx//2)
    print("nx_t:", nx_t)
    print("ny_t:", ny_t)

    kx_ours = (2 * math.pi) * nx_t / (Nx * dx)
    ky_ours = (2 * math.pi) * ny_t / (Ny * dx)

    # 3) 参考写法：与 FFT 频率严格对齐（fftfreq+fftshift）
    fx_ref = torch.fft.fftfreq(Nx, d=dx,dtype=dtype, device=device)
    fy_ref = torch.fft.fftfreq(Ny, d=dx,dtype=dtype, device=device)
    kx_ref = 2 * math.pi * torch.fft.fftshift(fx_ref)
    ky_ref = 2 * math.pi * torch.fft.fftshift(fy_ref)

    # 4) 计算差值（绝对差与以 Δk 归一的“格点差”）
    dkx = 2 * math.pi / (Nx * dx)
    dky = 2 * math.pi / (Ny * dx)
    max_abs_dx = torch.max(torch.abs(kx_ours - kx_ref)).item()
    max_abs_dy = torch.max(torch.abs(ky_ours - ky_ref)).item()
    max_bin_dx = torch.max(torch.abs((kx_ours - kx_ref) / dkx)).item()
    max_bin_dy = torch.max(torch.abs((ky_ours - ky_ref) / dky)).item()

    print("\n==== Off-by-one check (freq axis) ====")
    print(f"Δkx={dkx:.6e}, Δky={dky:.6e}")
    print(f"max|kx_ours-kx_ref| = {max_abs_dx:.6e}  (~ {max_bin_dx:.3f} bins)")
    print(f"max|ky_ours-ky_ref| = {max_abs_dy:.6e}  (~ {max_bin_dy:.3f} bins)")
    print(kx_ours)
    print(kx_ref)
    print(ky_ours)
    print(ky_ref)

    def verdict(val):
        if val < 1e-9:
            return "PASS (aligned)"
        # 接近 0.5 格为典型半格偏移
        if abs(val - 0.5) < 1e-2:
            return "WARN (≈ half-bin offset)"
        return "WARN (misaligned)"

    print("kx verdict:", verdict(max_bin_dx))
    print("ky verdict:", verdict(max_bin_dy))

    # 5) meshgrid 形状与等价性检查
    #   'xy' 与 MATLAB 一致：传入 (kx_1d, ky_1d) 得到形状 (Ny, Nx)
    #   'ij' 传入 (ky_1d, kx_1d) 也得到 (Ny, Nx)
    kx_xy, ky_xy = torch.meshgrid(kx_ours, ky_ours, indexing="xy")   # (Ny, Nx)
    ky_ij, kx_ij = torch.meshgrid(ky_ours, kx_ours, indexing="ij")   # (Ny, Nx)

    ksq_xy = (kx_xy ** 2 + ky_xy ** 2)  # (Ny, Nx)
    ksq_ij = (kx_ij ** 2 + ky_ij ** 2)  # (Ny, Nx)

    mesh_diff = torch.max(torch.abs(ksq_xy - ksq_ij)).item()

    print("\n==== meshgrid shape/alignment ====")
    print(f"k_xy shape: kx_xy={tuple(kx_xy.shape)}, ky_xy={tuple(ky_xy.shape)}")
    print(f"k_ij shape: kx_ij={tuple(kx_ij.shape)}, ky_ij={tuple(ky_ij.shape)}")
    print(f"max|k_sq(xy) - k_sq(ij)| = {mesh_diff:.6e}")
    print("meshgrid verdict:", "PASS" if mesh_diff < 1e-9 else "WARN (values differ)")

    return {
        "Nx": Nx,
        "Ny": Ny,
        "dx": dx,
        "max_bin_dx": max_bin_dx,
        "max_bin_dy": max_bin_dy,
        "mesh_diff": mesh_diff,
    }


# 运行检查（使用项目配置或默认值）
_ = check_off_by_one_and_meshgrid()
```

```python
#坐标和角谱法传播验证
import numpy as np
import matplotlib.pyplot as plt

# ------------------------
# 参数设置
# ------------------------
wavelength = 1550e-9     # 波长 (m)
k = 2 * np.pi / wavelength
Nx, Ny = 256, 512       # 采样点数
dx = dy = 8e-6          # 采样间隔 (m)
z = 0.01                # 传播距离 (m)

# ------------------------
# 构造空间坐标
# ------------------------

x = (np.arange(Nx)-Nx // 2) * dx
y = (np.arange(Ny)-Ny // 2) * dy
X, Y = np.meshgrid(x, y)

# ------------------------
# 输入光场：一个高斯波包
# ------------------------
w0 = 0.5e-3
E0 = np.exp(-(X**2 + Y**2) / w0**2)

# ------------------------
# 手动构造频率坐标 (kx, ky)
# ------------------------
fx = np.fft.fftfreq(Nx, d=dx)
fy = np.fft.fftfreq(Ny, d=dy)
FX, FY = np.meshgrid(fx, fy)

kx_ref = 2 * np.pi * np.fft.fftshift(FX)   # 正确参考形式
ky_ref = 2 * np.pi * np.fft.fftshift(FY)

# 你自己的 kx_ours （可修改为你的构造方式）
kx_ours = (np.arange(Nx)-Nx // 2) * (2 * np.pi) / (Nx * dx)
ky_ours = (np.arange(Ny)-Ny // 2) * (2 * np.pi) / (Ny * dy)
KX_ours, KY_ours = np.meshgrid(kx_ours, ky_ours)

# ------------------------
# 传播：角谱法
# ------------------------
def propagate(Ein, KX, KY):
    kz = np.sqrt(np.maximum(0, k**2 - KX**2 - KY**2))
    Ein_FT = np.fft.fftshift(np.fft.fft2(np.fft.ifftshift(Ein)))
    Eout_FT = Ein_FT * np.exp(1j * kz * z)
    Eout = np.fft.fftshift(np.fft.ifft2(np.fft.ifftshift(Eout_FT)))
    return Eout

E_ref = propagate(E0, kx_ref, ky_ref)
E_ours = propagate(E0, KX_ours, KY_ours)

# ------------------------
# 结果可视化
# ------------------------
plt.figure(figsize=(10,4))
plt.subplot(1,2,1)
plt.imshow(np.abs(E_ref)**2, cmap='jet')
plt.title('Ref Propagation')
plt.colorbar()

plt.subplot(1,2,2)
plt.imshow(np.abs(E_ours)**2, cmap='jet')
plt.title('Your kx_ours Propagation')
plt.colorbar()
plt.tight_layout()
plt.show()
```

```python
#滤波相位掩模
import torch
import numpy as np
import matplotlib.pyplot as plt

# ------------------------
# 参数设置
# ------------------------
Nx, Ny = 256, 512      # 例如 Ny 与 Nx 不等
dx = dy = 8e-6         # 采样间隔 (m)
device = "cpu"

# 构造一个随机相位掩模（模拟 SLM 相位）
torch.manual_seed(0)
phase_mask = torch.rand((Ny, Nx), dtype=torch.float32, device=device) * 2 * np.pi - np.pi

# 超高斯参数
d0_pixels = 0.75*150              # d0 用像素为单位（水平方向宽度）
order = 4                    # 超高斯阶数
d0y_pixels = d0_pixels * (Ny / Nx)   # 竖向按 Ny/Nx 缩放




def apply_frequency_filtering(mask: torch.Tensor, d0: float, aspect_ratio: float = 2.0) -> torch.Tensor:
    """
    应用频域超高斯滤波平滑相位掩模。
    
    原理：通过 FFT 变换到频域，乘以低通滤波器去除高频噪声，再 IFFT 回空域。
    这样可以让相位分布更平滑，减少散射损耗，扩展工作带宽。
    
    参数:
        mask: (Ny, Nx) 相位掩模张量，范围 [-π, π]
        d0: 频域滤波器特征宽度（像素），越小越平滑，但性能可能下降
        aspect_ratio: 滤波器长宽比（垂直/水平），用于匹配空域的反向特性
                     空域竖长(Ny>Nx)时，频域应横长，需 aspect_ratio<1
    
    返回:
        平滑后的相位掩模，形状与输入相同
    """
    Ny, Nx = mask.shape
    device = mask.device
    
    # 1. FFT 变换到频域
    mask_fft = torch.fft.fft2(mask)
    mask_fft_shifted = torch.fft.fftshift(mask_fft)
    
     # 构造像素空间的频率索引（以像素为单位），中心化
    # range: -Nx/2 ... Nx/2-1
    fx_idx = (torch.arange(Nx, device=mask.device, dtype=torch.float32) - Nx//2)
    fy_idx = (torch.arange(Ny, device=mask.device, dtype=torch.float32) - Ny//2)
    FX, FY = torch.meshgrid(fx_idx, fy_idx, indexing='xy')  # shape (Ny,Nx)
    
    # 4阶超高斯：exp(-((fx/fx0)^2 + (fy/fy0)^2)^4)
    # 比普通高斯更"平顶"，过渡区更陡峭
    # 注意：空域窄的方向，频域应该宽（反向关系）
    fy_cutoff = d0 * aspect_ratio  # 垂直方向
    fx_cutoff = d0                  # 水平方向（基准）
    filter_freq = torch.exp(-((FX / fx_cutoff)**2 + (FY / fy_cutoff)**2)**4)
    
    # 3. 频域相乘（低通滤波）
    mask_filtered_fft = mask_fft_shifted * filter_freq
    
    # 4. IFFT 变回空域
    mask_smooth = torch.fft.ifft2(torch.fft.ifftshift(mask_filtered_fft)).real

    return mask_smooth, filter_freq, mask_fft_shifted, mask_filtered_fft

# ------------------------
# 应用两个滤波器
# ------------------------
mask_pixel, H_pixel, Fshift_pixel, Ffilt_pixel = apply_frequency_filtering(phase_mask, d0_pixels,2.0)



# ------------------------
# 可视化：空间域、滤波器、以及四个频域图
# ------------------------
plt.figure(figsize=(14, 10))

# 空间域对比
plt.subplot(3, 4, 1)
plt.imshow(phase_mask.cpu(), cmap='twilight', vmin=-np.pi, vmax=np.pi, aspect='auto')
plt.title("Original Phase Mask")
plt.colorbar(fraction=0.045)
plt.axis('image')
plt.subplot(3, 4, 2)
plt.imshow(mask_pixel.cpu(), cmap='twilight', vmin=-np.pi, vmax=np.pi, aspect='auto')
plt.title("Pixel-Frequency Filtered (space)")
plt.colorbar(fraction=0.045)
plt.axis('image')

# 滤波器形状
plt.subplot(3, 4, 4)
plt.imshow(H_pixel.cpu(), cmap='viridis', aspect='auto')
plt.title(f"Pixel Filter (d0={d0_pixels:.1f}px, order={order})")
plt.colorbar(fraction=0.045)
plt.axis('image')


# 频域对比（像素滤波器：shifted before, shifted after）
plt.subplot(3, 4, 6)
plt.imshow(torch.log1p(torch.abs(Fshift_pixel)).cpu(), cmap='inferno', aspect='auto')
plt.title("|mask_fft_shifted| (before pixel filter)")
plt.colorbar(fraction=0.045)
plt.axis('image')
plt.subplot(3, 4, 7)
plt.imshow(torch.log1p(torch.abs(Ffilt_pixel)).cpu(), cmap='inferno', aspect='auto')
plt.title("|mask_fft_filtered| (after pixel filter)")
plt.colorbar(fraction=0.045)
plt.axis('image')


# 中心剖面对比（横向）
center_row = Ny // 2
plt.subplot(3, 4, 10)
plt.plot(phase_mask.cpu()[center_row, :].numpy(), label='orig')
plt.plot(mask_pixel.cpu()[center_row, :].numpy(), label='pixel')

plt.legend()
plt.title("Center row profile (space)")

# 中心频谱剖面（对比）
cx = Nx // 2
plt.subplot(3, 4, 11)
plt.plot(torch.log1p(torch.abs(Fshift_pixel)).cpu()[center_row, :].numpy(), label='fft_shift_before_pixel')
plt.plot(torch.log1p(torch.abs(Ffilt_pixel)).cpu()[center_row, :].numpy(), label='fft_shift_after_pixel')
plt.legend()
plt.title("Center freq profile (pixel)")


plt.tight_layout()
plt.show()
```
